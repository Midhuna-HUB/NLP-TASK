```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk
from nltk.probability import FreqDist
from nltk.util import bigrams, trigrams
from nltk.classify import NaiveBayesClassifier
from nltk.classify.util import accuracy
from nltk.corpus import wordnet as wn

# =========================
# DOWNLOAD REQUIRED DATA
# =========================
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('wordnet')

# =========================
# INPUT
# =========================
text = input("Enter News Article:\n")

# =========================
# TOKENIZATION
# =========================
tokens = word_tokenize(text)

# =========================
# POS TAGGING
# =========================
tagged = pos_tag(tokens)
print("\n--- POS TAGS ---")
print(tagged)

# =========================
# SIMPLE PARSING (SVO)
# =========================
subject = ""
verb = ""
obj = ""

for word, tag in tagged:
    if tag.startswith("NN") and subject == "":
        subject = word
    elif tag.startswith("VB") and verb == "":
        verb = word
    elif tag.startswith("NN") and subject != "" and verb != "":
        obj = word
        break

print("\n--- PARSING ---")
print("Subject:", subject)
print("Verb:", verb)
print("Object:", obj)

# =========================
# NAMED ENTITY RECOGNITION
# =========================
print("\n--- NAMED ENTITIES ---")
entities = ne_chunk(tagged)
print(entities)

# =========================
# WORD FREQUENCY
# =========================
words = [w.lower() for w in tokens if w.isalpha()]
fdist = FreqDist(words)

print("\n--- WORD FREQUENCY ---")
for word, count in fdist.most_common(10):
    print(word, ":", count)

# =========================
# PROBABILITY
# =========================
total_words = len(words)
print("\n--- PROBABILITY ---")
for word, count in fdist.most_common(5):
    print(f"P({word}) = {count}/{total_words} = {round(count/total_words, 2)}")

# =========================
# BIGRAMS
# =========================
print("\n--- BIGRAMS ---")
print(list(bigrams(words)))

# =========================
# TRIGRAMS
# =========================
print("\n--- TRIGRAMS ---")
print(list(trigrams(words)))

# =========================
# NAIVE BAYES CLASSIFICATION
# =========================
train_data = [
    ({"excellent": True, "good": True}, "Positive"),
    ({"friendly": True}, "Positive"),
    ({"bad": True, "poor": True}, "Negative"),
    ({"terrible": True}, "Negative")
]

classifier = NaiveBayesClassifier.train(train_data)

features = {
    "excellent": "excellent" in text.lower(),
    "good": "good" in text.lower(),
    "friendly": "friendly" in text.lower(),
    "bad": "bad" in text.lower(),
    "poor": "poor" in text.lower(),
    "terrible": "terrible" in text.lower()
}

category = classifier.classify(features)
print("\n--- CLASSIFICATION ---")
print("Category:", category)

# =========================
# ACCURACY
# =========================
test_data = [
    ({"excellent": True, "good": True}, "Positive"),
    ({"bad": True, "poor": True}, "Negative")
]

print("\n--- ACCURACY ---")
print("Accuracy:", accuracy(classifier, test_data))

# =========================
# SEMANTIC ANALYSIS (WORDNET)
# =========================
print("\n--- SEMANTIC MEANING (WORDNET) ---")
for syn in wn.synsets("technology")[:3]:
    print("-", syn.definition())
```
