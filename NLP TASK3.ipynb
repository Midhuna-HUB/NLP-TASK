```python

import nltk
import string
import matplotlib.pyplot as plt
from collections import Counter

# Download stopwords
nltk.download('stopwords')
from nltk.corpus import stopwords


articles = [
    "Artificial intelligence is transforming healthcare and education systems",
    "AI and machine learning are the future of technology and innovation",
    "Healthcare systems are improving with artificial intelligence solutions",
    "Education technology is growing rapidly with AI based tools"
]


stop_words = set(stopwords.words('english'))


cleaned_words = []

for text in articles:
    text = text.lower()  # lowercase

    # remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # tokenize
    words = text.split()

    # remove stopwords
    filtered_words = [word for word in words if word not in stop_words]

    cleaned_words.extend(filtered_words)


freq = Counter(cleaned_words)
print("\nWord Frequency:\n", freq)


vocab = set(cleaned_words)
print("\nVocabulary:\n", vocab)
print("Vocabulary Size:", len(vocab))


top5 = freq.most_common(5)
print("\nTop 5 Words:", top5)


total_tokens = len(cleaned_words)
unique_tokens = len(vocab)

ttr = unique_tokens / total_tokens
print("\nType Token Ratio (TTR):", round(ttr, 3))


tech_keywords = [
    "ai", "artificial", "intelligence",
    "technology", "machine", "learning"
]

tech_count = sum(freq[word] for word in tech_keywords)

print("\nTech-related word count:", tech_count)

if tech_count > 0:
    print("Dominant Topic: Technology & AI")


words = [w[0] for w in top5]
counts = [w[1] for w in top5]

plt.figure(figsize=(6,4))
plt.bar(words, counts)
plt.title("Top 5 Word Frequency")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.show()
```
