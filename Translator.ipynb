```python
# ============================================================
# ENGLISH → TAMIL AI TRANSLATOR
# Pretrained Transformer + PyTorch
# Interactive Translator + Dataset Testing
# ============================================================

import os
import re
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM


# ============================================================
# 1. SETTINGS
# ============================================================

MODEL_NAME = "Varsha00/finetuned-opusmt-en-to-ta"

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("=" * 60)
print("ENGLISH → TAMIL AI TRANSLATOR")
print("=" * 60)
print("Device:", DEVICE)
print("Model:", MODEL_NAME)


# ============================================================
# 2. LOAD PRETRAINED MODEL
# ============================================================

print("\nLoading tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("Tokenizer loaded successfully.")

print("\nLoading translation model...")

model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

model = model.to(DEVICE)

model.eval()

print("Translation model loaded successfully.")


# ============================================================
# 3. TEXT CLEANING
# ============================================================

def clean_text(text):

    text = str(text).strip()

    text = re.sub(r"\s+", " ", text)

    return text


# ============================================================
# 4. TRANSLATION FUNCTION
# ============================================================

def translate_sentence(sentence):

    sentence = clean_text(sentence)

    if not sentence:

        return ""

    inputs = tokenizer(
        sentence,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=128
    )

    inputs = {
        key: value.to(DEVICE)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        translated = model.generate(
            **inputs,
            max_length=128,
            num_beams=5,
            early_stopping=True
        )

    result = tokenizer.decode(
        translated[0],
        skip_special_tokens=True
    )

    return result.strip()


# ============================================================
# 5. TEST TRANSLATIONS
# ============================================================

test_sentences = [

    "How are you?",
    "What is your name?",
    "Good morning.",
    "Thank you.",
    "Where are you going?",
    "I am a student.",
    "I like music.",
    "I am happy.",
    "I need water.",
    "Can you help me?",
    "I am learning Python.",
    "I am studying computer science.",
    "I want to learn artificial intelligence.",
    "Where is the hospital?",
    "I am going to college.",
    "Please help me.",
    "I don't understand.",
    "See you tomorrow.",
    "Have a nice day.",
    "Everything is fine."
]


print("\n")
print("=" * 60)
print("TEST TRANSLATIONS")
print("=" * 60)


for sentence in test_sentences:

    translation = translate_sentence(sentence)

    print("\nEnglish :", sentence)
    print("Tamil   :", translation)
    print("-" * 60)


# ============================================================
# 6. LOAD YOUR 110-PAIR DATASET
# ============================================================

DATASET_PATH = "english_tamil_parallel_corpus.csv"


if os.path.exists(DATASET_PATH):

    print("\n")
    print("=" * 60)
    print("YOUR DATASET")
    print("=" * 60)

    df = pd.read_csv(DATASET_PATH)

    print("\nDataset loaded successfully.")

    print("Number of pairs:", len(df))

    print("\nColumns:")

    print(df.columns.tolist())


    # Check required columns

    if "english" in df.columns and "tamil" in df.columns:

        print("\nTesting your dataset...")

        correct_examples = 0

        total_examples = min(20, len(df))


        for i in range(total_examples):

            english_text = str(df.iloc[i]["english"])

            expected_tamil = str(df.iloc[i]["tamil"])

            predicted_tamil = translate_sentence(
                english_text
            )

            print("\nEnglish :", english_text)

            print("Expected:", expected_tamil)

            print("Model   :", predicted_tamil)

            print("-" * 60)

            correct_examples += 1


        print("\nDataset testing completed.")

    else:

        print(
            "\nCSV must contain columns named "
            "'english' and 'tamil'."
        )


else:

    print("\nCSV file not found.")

    print(
        "The pretrained translator will still work."
    )


# ============================================================
# 7. SIMPLE TRANSLATION FUNCTION FOR NOTEBOOK
# ============================================================

print("\n")
print("=" * 60)
print("READY FOR TRANSLATION")
print("=" * 60)

print(
    "\nUse the following command to translate:"
)

print(
    'translate_sentence("How are you?")'
)


# ============================================================
# 8. INTERACTIVE TRANSLATOR
# ============================================================

print("\n")
print("=" * 60)
print("INTERACTIVE ENGLISH → TAMIL TRANSLATOR")
print("=" * 60)

print("Type an English sentence.")
print("Type 'exit' to stop.")
print("=" * 60)


while True:

    user_sentence = input(
        "\nEnter English sentence: "
    )

    if user_sentence.lower().strip() == "exit":

        print("\nTranslator stopped.")

        break


    if not user_sentence.strip():

        print(
            "Please enter an English sentence."
        )

        continue


    translation = translate_sentence(
        user_sentence
    )


    print(
        "\nTamil Translation:",
        translation
    )
```
